```python
import pandas as pd
from scipy.io import arff
from sklearn.feature_selection import f_classif
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset 'column_diagnosis´
data = arff.loadarff('column_diagnosis.arff')
df = pd.DataFrame(data[0])

# Extract features (X) and target variable (y)
X = df.iloc[:, :-1].values 
y = df.iloc[:, -1].astype('category').cat.codes.values

# Apply f_classif
f_values, p_values = f_classif(X, y)

highest_discriminative_power_index = f_values.argmax()
lowest_discriminative_power_index = f_values.argmin()

# Define a function for plotting class-conditional PDFs
def plot_pdf(feature_index):
    plt.figure(figsize=(10, 6))
    for label in df['class'].unique():
        sns.kdeplot(df[df['class'] == label].iloc[:, feature_index], label=f'Class {label}')
    plt.title(f'Class-Conditional PDFs for Feature {feature_index}')
    plt.legend()
    plt.show()

# Plot PDFs for the feature with highest discriminative power
plot_pdf(highest_discriminative_power_index)

# Plot PDFs for the feature with lowest discriminative power
plot_pdf(lowest_discriminative_power_index)


```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

# Load the data (assuming you've already loaded the data into a variable `X` for features and `y` for target)
# If not, load the data using the appropriate method.

# Initialize lists to store training and testing accuracies
train_accuracies = []
test_accuracies = []

# Define the depth limits
depth_limits = [1, 2, 3, 4, 5, 6, 8, 10]

# Set the seed for reproducibility
np.random.seed(0)

# Perform 10 runs for each depth limit
for depth in depth_limits:
    train_accuracy = 0
    test_accuracy = 0
    
    for _ in range(10):
        # Split the data into training and testing sets
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=0)
        
        # Train a decision tree classifier with the specified depth limit
        clf = DecisionTreeClassifier(max_depth=depth, random_state=0)
        clf.fit(X_train, y_train)
        
        # Calculate training and testing accuracies
        train_accuracy += clf.score(X_train, y_train)
        test_accuracy += clf.score(X_test, y_test)
    
    # Average the accuracies over 10 runs
    train_accuracies.append(train_accuracy / 10)
    test_accuracies.append(test_accuracy / 10)

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(depth_limits, train_accuracies, label='Training Accuracy', marker='o')
plt.plot(depth_limits, test_accuracies, label='Testing Accuracy', marker='o')
plt.xlabel('Depth Limit')
plt.ylabel('Accuracy')
plt.title('Training and Testing Accuracies vs. Depth Limit')
plt.legend()
plt.grid(True)
plt.show()
```

```python
´´´
Depth Limit and Overfitting:

As the depth limit increases, the training accuracy steadily improves. This is expected because a deeper tree can capture more complex relationships in the data.
However, the testing accuracy starts to plateau and may even decrease after a certain depth limit. This is a clear sign of overfitting, where the model learns noise in the training data that does not generalize well to unseen data.
Generalization Capacity:

The point at which the testing accuracy plateaus or starts to decline indicates the model's generalization capacity. In this case, it suggests that a depth limit of around 3 to 5 provides the best balance between capturing important patterns and avoiding overfitting.
High Depth Limits:

Depth limits beyond this optimal range lead to overfitting. The model becomes excessively complex, fitting to noise and idiosyncrasies in the training data that do not hold in the broader population.
Underfitting:

When the depth limit is very low (e.g., 1 or 2), both training and testing accuracies are relatively low. This is a sign of underfitting, indicating that the model is too simple to capture the underlying patterns in the data.
Generalization and Data Robustness:

The results also imply that the model's performance is robust across different settings (due to the 10 runs per parameterization). This means that the model's generalization capacity is consistent and not overly sensitive to random variations in the training and testing sets.
Final Model Selection:

Based on these observations, it is advisable to select a depth limit in the range of 3 to 5 for the final decision tree model. This choice strikes a balance between capturing relevant information and avoiding overfitting, resulting in a model with good generalization capacity.
´´´
```

```python
from sklearn.tree import plot_tree

# Convert class labels to strings and then to a list
class_names = list(df['class'].unique().astype(str))

# Initialize a Decision Tree Classifier with specified parameters
clf = DecisionTreeClassifier(random_state=0, min_samples_leaf=20)

# Train the decision tree using all available data
clf.fit(X, y)

# i. Plot the decision tree
plt.figure(figsize=(12, 8))
plot_tree(clf, filled=True, class_names=class_names, rounded=True, fontsize=10)
plt.show()
```
